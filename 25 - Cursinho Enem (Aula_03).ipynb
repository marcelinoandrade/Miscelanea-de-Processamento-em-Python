- **Conteúdos associados:** **funções logarítmicas**, **modelagem de fenômenos naturais**, **interpretação de grandezas físicas**.  

**Modelagem e solução resumida:**  
1) Dados:  
   $
   M_2 - M_1 = \frac{2}{3}\log\left(\frac{E_2}{E_1}\right)
   $
   Sabendo que $M_1 = 6{,}9$ e $E_1 = \frac{1}{10}$ e $E_2 \Rightarrow \frac{E_2}{E_1} = 10$.  

2) Substituindo:  
   $
   M_2 - 6{,}9 = \frac{2}{3}\log(10)
   $


   $
   M_2 - 6{,}9 = \frac{2}{3} \times1 
   $

   >>
   $
   M_2 \approx 7{,}6
   $

**Resposta:** **Alternativa C (7,6).**



```python
# Q150 — Escala Richter: magnitude vs energia (ENEM 2024)
# Regras: matplotlib, um gráfico por figura, sem definir cores específicas.

import numpy as np
import matplotlib.pyplot as plt

# Dados do enunciado
M1 = 6.9           # magnitude do 1º terremoto
ratio = 10.0       # E2/E1 (energia do 2º é 10× a do 1º)

# Modelo (escala Richter)
# M2 - M1 = (2/3) * log10(E2/E1)
deltaM = (2/3) * np.log10(ratio)
M2 = M1 + deltaM

print(f"M1 = {M1}")
print(f"E2/E1 = {ratio:.1f}")
print(f"M2 = {M2:.1f}")  # esperado ~ 7.6

# Curva para diferentes razões de energia (para fixação do modelo)
r = np.logspace(-2, 3, 400)          # E2/E1 variando de 0,01 a 1000
M2_curve = M1 + (2/3) * np.log10(r)

# Gráfico
plt.figure(figsize=(8,4))
plt.plot(r, M2_curve, linewidth=1.5, label="M2 para diferentes E2/E1")
plt.scatter([ratio], [M2], s=40, label="Caso da questão")
plt.xscale("log")
plt.xlabel("Razão de energias E2/E1 (escala log)")
plt.ylabel("Magnitude M2")
plt.title("Q150 – Escala Richter: M2 em função de E2/E1 (M1 = 6,9)")
plt.grid(True, linestyle=":", linewidth=0.6)
plt.legend()
plt.tight_layout()
plt.show()

# Opcional: salvar figura
# plt.savefig("q150_richter.png", dpi=200)
```

- **H31 – Funções e modelos matemáticos em contextos reais:** usa **reta de tendência** para projetar receita futura.  
- **Conteúdos:** **função afim (modelo linear)**, **interpretação de gráfico**, **projeção econômica**.

**Modelagem e solução resumida:**  
1) Pegue dois pontos da **reta de tendência** no gráfico: $(2014,\;3{,}5)$ e $(2021,\;6{,}5)$ milhões.  
2) Inclinação: $m=\dfrac{6{,}0-4{,}0}{2020-2014}=\dfrac{2}{6}\approx0{,}33$ mi/ano.  
3) Projeção para 2026: $6{,}0 + 6\cdot0{,}33 \approx 8{,}0$ milhões.

**Resposta:** **Alternativa B (8).**


```python
# Q152 — Projeção por reta de tendência (ENEM 2024)
# Regra: matplotlib, um gráfico por figura, sem definir cores específicas.

import numpy as np
import matplotlib.pyplot as plt

# Dois pontos da reta de tendência lidos do gráfico
x0, y0 = 2014, 4.0   # (ano, receita em milhões)
x1, y1 = 2020, 6.0

# Coeficientes da função afim: y = m*x + b
m = (y1 - y0) / (x1 - x0)
b = y0 - m * x0

# Projeção para 2026
x_proj = 2026
y_proj = m * x_proj + b

print(f"Inclinação m = {m:.3f} mi/ano")
print(f"Receita estimada em {x_proj}: {y_proj:.2f} milhões ≈ {round(y_proj):.0f} milhões")

# Curva da tendência de 2014 a 2026
anos = np.arange(2014, 2027)
receitas = m * anos + b

# Gráfico
plt.figure(figsize=(10,4))
plt.plot(anos, receitas, linewidth=1.8, label="Reta de tendência")
plt.scatter([x0, x1], [y0, y1], s=40, label="Pontos usados da reta")
plt.scatter([x_proj], [y_proj], s=60, marker="o", label=f"Estimativa {x_proj} ≈ {round(y_proj):.0f} mi")
plt.xlabel("Ano")
plt.ylabel("Receita anual (milhões de reais)")
plt.title("Q152 – Estimativa por tendência linear")
plt.grid(True, linestyle=":", linewidth=0.6)
plt.legend()
plt.tight_layout()
plt.show()

# Opcional: salvar a figura
# plt.savefig("q152_tendencia_linear.png", dpi=200)
```

**Enquadramento (por que é C8):**  
- **H30/H31 – Crescimento exponencial em contexto real:** dimensiona o **momento de aplicação** de um produto a partir de um **modelo exponencial** de reprodução.  
- **Conteúdos:** **função exponencial**, **PG**, **modelagem em manejo agrícola**.

**Modelagem e solução resumida:**  
- População multiplica por **10 a cada 3 dias**. Início: $N_0=100$.  
- Dose: **1 L / 200\,000** larvas. Produto total: **5 L** ⇒ atende **$5\times200\,000=1\,000\,000$** larvas.  
- Procurar \(t\) tal que \(N(t)=1\,000\,000\):  
  $
  N(t)=100\cdot 10^{t/3}=1\,000\,000
  \Rightarrow 10^{t/3}=10^{4}\Rightarrow \frac{t}{3}=4\Rightarrow t=12\text{ dias.}
  $

**Resposta:** **12 dias (Alternativa D).**


```python
# Q156 — Crescimento exponencial da população de larvas (ENEM 2024)
# Regras: matplotlib, um gráfico por figura, sem definir cores específicas.

import numpy as np
import matplotlib.pyplot as plt

# Dados do problema
N0 = 100                 # população inicial
fator = 10               # multiplica por 10 a cada 3 dias
dias_por_ciclo = 3
produto_total = 5        # litros
dose_por_200k = 1        # litro por 200.000 larvas

# População máxima que o produto cobre
Nmax = produto_total * 200_000  # 1.000.000 larvas

# Modelo exponencial: N(t) = N0 * fator^(t/3)
t = np.linspace(0, 18, 100)
N = N0 * fator**(t / dias_por_ciclo)

# Encontrar o ponto em que N(t) = Nmax
t_aplicacao = 3 * np.log10(Nmax / N0)
N_aplicacao = N0 * fator**(t_aplicacao / dias_por_ciclo)

print(f"Aplicar o produto após {t_aplicacao:.0f} dias (N ≈ {N_aplicacao:.0f} larvas).")

# Gráfico
plt.figure(figsize=(8,4))
plt.plot(t, N, linewidth=2, label=r"$N(t)=100\cdot10^{t/3}$")
plt.axhline(Nmax, linestyle="--", linewidth=1.2, label="Limite para aplicação (1.000.000 larvas)")
plt.scatter([t_aplicacao], [N_aplicacao], s=60, label=f"Aplicar aos {t_aplicacao:.0f} dias")
plt.yscale("log")
plt.xlabel("Tempo (dias)")
plt.ylabel("População de larvas (escala logarítmica)")
plt.title("Q156 – Crescimento exponencial das larvas")
plt.grid(True, linestyle=":", linewidth=0.6)
plt.legend()
plt.tight_layout()
plt.show()

# Opcional: salvar a figura
# plt.savefig("q156_crescimento_larvas.png", dpi=200)
```

**Enquadramento (por que é C8):**  
- **H31 – Funções e modelos matemáticos em contextos reais:** aplica uma **função logarítmica** para modelar o nível sonoro (decibel) em função da **intensidade física do som**.  
- **Conteúdos:** **função logarítmica**, **escala em decibéis**, **modelagem física de intensidade sonora**.

**Modelagem e solução resumida:**  
A relação entre nível sonoro e intensidade é dada por:  
$
N = 10\log_{10}\left(\frac{I}{I_0}\right)
$  
Com $I = 8\times10^{-8}$ e $I_0 = 10^{-12}$:  
$
N = 10\log_{10}(8\times10^4) = 10(\log_{10}8 + 4)
$  
Sabendo que $\log_{10}8 = 3\log_{10}2 \approx 3\times0{,}3 = 0{,}9$:  
$
N = 10(0{,}9 + 4) = 49 \text{ dB}
$

**Classificação:**  
$48 < N < 55$ → **som moderado** → **Medida preventiva II**.

**Resposta:** **Alternativa B (II).**


```python
# Q163 — Escala de decibéis com faixas e legenda lateral (ENEM 2024)
# Destaque visual com legenda à direita das faixas

import numpy as np
import matplotlib.pyplot as plt

# Relação N = 10·log10(I/I0)
I0 = 1e-12
I = np.logspace(-12, -6, 500)
N = 10 * np.log10(I / I0)

# Faixas (dB): (min, max, rótulo, cor)
faixas = [
    (0, 48,  "I • som baixo",        "#CDE7FF"),
    (48, 55, "II • som moderado",    "#B0D6FF"),
    (55, 80, "III • moderado alto",  "#FFE58A"),
    (80, 115,"IV • som alto",        "#FFB085"),
    (115,130,"V • som muito alto",   "#FF8A8A"),
]

# Nova máquina
I_new = 8e-8
N_new = 10 * np.log10(I_new / I0)

plt.figure(figsize=(10,5))
plt.plot(N, I, color="black", linewidth=2.2, label=r"$N=10\log_{10}(I/I_0)$")
plt.xlim(0,130)
plt.ylim(1e-12,1e-6)
plt.yscale("log")
plt.grid(True, linestyle=":", linewidth=0.6)

# Desenha as faixas coloridas
for nmin, nmax, rot, cor in faixas:
    plt.axvspan(nmin, nmax, color=cor, alpha=0.55)

# Ponto da nova máquina
plt.scatter([N_new], [I_new], s=100, color="red", edgecolor="black", zorder=5,
            label=f"Máquina: {N_new:.1f} dB")

# Texto lateral de legenda (ao lado direito)
x_leg = 133
y_pos = np.geomspace(1e-6, 1e-12, len(faixas))  # posições verticais
for (nmin, nmax, rot, cor), y in zip(faixas, y_pos):
    plt.text(x_leg, y, rot, fontsize=10, fontweight="bold",
             color="black", va="center", ha="left",
             bbox=dict(facecolor=cor, alpha=0.6, edgecolor="none", boxstyle="round,pad=0.3"))

plt.xlabel("Nível sonoro (dB)")
plt.ylabel("Intensidade (W/m²)")
plt.title("Q163 – Escala logarítmica de intensidade sonora e medidas preventivas")
plt.legend(loc="lower right", fontsize=9)
plt.tight_layout()
plt.show()

# Opcional: salvar imagem
# plt.savefig("q163_legenda_lateral.png", dpi=200, bbox_inches="tight")
```

**Enquadramento (por que é C8):**  
- **H31 – Funções e modelos matemáticos em contextos reais:** determina **taxa mensal de juros compostos** a partir de um **fluxo de caixa** equivalente ao preço à vista.  
- **Conteúdos:** **juros compostos (função exponencial discreta)**, **equivalência de capitais**, **equação quadrática**.

**Modelagem e solução (usando PV, P e série geométrica):**
Significados:

  $ PV $: valor à vista (ou valor presente)
  
  $ P $: valor de cada parcela (fixa e periódica)
  
  $ i $: taxa de juros por período (em forma decimal)
  
  $ n $: número de parcelas


- **Dados:** preço à vista $PV=1\,500$ Financiado com +20% → $1{,}2\cdot1500=1\,800$ em **3 parcelas iguais** ⇒
  $
  P=\frac{1\,800}{3}=600.
  $

- **Equações (Tabela Price):**
  $
  \boxed{PV=P\cdot\frac{1-(1+i)^{-n}}{i}}
  \qquad\Longleftrightarrow\qquad
  \boxed{P=\frac{PV\cdot i}{1-(1+i)^{-n}}}
  $
  (derivadas de $PV=\sum_{k=1}^{n}\frac{P}{(1+i)^k}$, soma de **PG**.)

- **Aplicação com \(n=3\):**
  $
  1500=600\cdot\frac{1-(1+i)^{-3}}{i}.
  $
  Pela **série geométrica**:
  $
  PV=\sum_{k=0}^{2}\frac{600}{(1+i)^k}
  \;\Rightarrow\;
  600+\frac{600}{1+i}+\frac{600}{(1+i)^2}=1500.
  $

- **Substituição $x=\frac{1}{1+i}$:**
 $
  600(1+x+x^2)=1500\;\Rightarrow\;x^2+x-1{,}5=0.
  $
  $
  x=\frac{-1+\sqrt{7}}{2}=\frac{-1+\frac{\sqrt{28}}{2}}{2}\approx0{,}8225
  \quad(\sqrt{28}\approx5{,}29).
  $

- **Taxa:**
  $
  i=\frac{1}{x}-1\approx\frac{1}{0{,}8225}-1\approx0{,}215
  \;\Rightarrow\; \boxed{21{,}5\%\ \text{ao mês}}.
  $

**Resposta:** alternativa **D (21,5%)**.


```python
# ENEM Q170 — Financiamento com 3 parcelas fixas (R$ 600) e juros compostos de 21,5% a.m.
# Agora com dois gráficos separados: composição das parcelas e saldo devedor

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Dados do problema
PV = 1500.0       # valor à vista (equivalente)
P  = 600.0        # valor da parcela fixa
i  = 0.215        # taxa de juros ao mês (21,5%)
n  = 3            # número de parcelas

# Lista para armazenar as linhas da tabela
linhas = []

# t = 0 (primeira parcela paga no ato da compra)
saldo = PV
juros = 0.0
amort = P
saldo -= amort
linhas.append([0, P, juros, amort, saldo])

# t = 1 e 2 (parcelas com incidência de juros)
for k in range(1, n):
    juros = saldo * i
    amort = P - juros
    saldo -= amort
    linhas.append([k, P, juros, amort, saldo if saldo > 0 else 0])

# Criar DataFrame
df = pd.DataFrame(linhas, columns=["Parcela (t)", "Parcela (R$)", "Juros (R$)", "Amortização (R$)", "Saldo Devedor (R$)"])
display(df.round(2))

# ------------------- GRÁFICO 1 -------------------
plt.figure(figsize=(8, 4))
plt.plot(df["Parcela (t)"], df["Juros (R$)"], 'r-o', linewidth=2, label="Juros (R$)")
plt.plot(df["Parcela (t)"], df["Amortização (R$)"], 'g-o', linewidth=2, label="Amortização (R$)")
plt.plot(df["Parcela (t)"], df["Parcela (R$)"], 'b--', linewidth=1.5, label="Parcela total (R$600)")
plt.title("Q170 – Composição das Parcelas\nTaxa de juros: 21,5% ao mês (juros compostos)", fontsize=12, fontweight="bold")
plt.xlabel("Número da Parcela (t)")
plt.ylabel("Valor (R$)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

# ------------------- GRÁFICO 2 -------------------
plt.figure(figsize=(8, 4))
plt.plot(df["Parcela (t)"], df["Saldo Devedor (R$)"], 'k-s', linewidth=2.5, label="Saldo devedor (R$)")
plt.title("Q170 – Evolução do Saldo Devedor\nTaxa de juros: 21,5% ao mês", fontsize=12, fontweight="bold")
plt.xlabel("Número da Parcela (t)")
plt.ylabel("Saldo Devedor (R$)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

# Totais
print(f"Total pago: R$ {df['Parcela (R$)'].sum():.2f}")
print(f"Total de juros pagos: R$ {df['Juros (R$)'].sum():.2f}")
print(f"Total amortizado: R$ {df['Amortização (R$)'].sum():.2f}")
print(f"Saldo final: R$ {df.iloc[-1]['Saldo Devedor (R$)']:.2f}")
```

```python
# Simulação de crédito com entrada igual à parcela fixa — Tabela Price
# Juros compostos de 1,5% ao mês (≈ 20% ao ano)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Dados do financiamento
PV = 1500.0       # valor à vista (ou preço equivalente)
i  = 0.015        # taxa de juros mensal (1,5%)
n  = 12           # número de parcelas após a entrada

# Queremos parcelas fixas, sendo a primeira paga no ato (mês 0)
# Portanto, valor presente é a soma dos valores descontados de cada parcela
# PV = P + P/(1+i) + P/(1+i)^2 + ... + P/(1+i)^(n-1)

# Cálculo da parcela P
P = PV / sum((1 / (1 + i)**k) for k in range(n))

# Montagem da tabela de amortização
saldo = PV
linhas = []

# Mês 0 – entrada (primeira parcela)
juros = 0.0
amort = P
saldo -= amort
linhas.append([0, P, juros, amort, saldo])

# Meses 1 a 11 (restantes)
for k in range(1, n):
    juros = saldo * i
    amort = P - juros
    saldo -= amort
    linhas.append([k, P, juros, amort, saldo if saldo > 0 else 0])

df = pd.DataFrame(linhas, columns=["Parcela (t)", "Pagamento (R$)", "Juros (R$)", "Amortização (R$)", "Saldo Devedor (R$)"])
display(df.round(2))

# ------------------- GRÁFICO 1 -------------------
plt.figure(figsize=(8, 4))
plt.plot(df["Parcela (t)"], df["Juros (R$)"], 'r-o', linewidth=2, label="Juros (R$)")
plt.plot(df["Parcela (t)"], df["Amortização (R$)"], 'g-o', linewidth=2, label="Amortização (R$)")
plt.plot(df["Parcela (t)"], df["Pagamento (R$)"], 'b--', linewidth=1.5, label=f"Parcela total fixa (R${P:.2f})")
plt.title("Crédito com entrada igual à parcela fixa\nTaxa: 1,5% ao mês (≈ 20% ao ano)", fontsize=12, fontweight="bold")
plt.xlabel("Número da Parcela (t)")
plt.ylabel("Valor (R$)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

# ------------------- GRÁFICO 2 -------------------
plt.figure(figsize=(8, 4))
plt.plot(df["Parcela (t)"], df["Saldo Devedor (R$)"], 'k-s', linewidth=2.5, label="Saldo devedor (R$)")
plt.title("Evolução do saldo devedor — Entrada igual à parcela fixa\nTaxa: 1,5% ao mês (Tabela Price)", fontsize=12, fontweight="bold")
plt.xlabel("Número da Parcela (t)")
plt.ylabel("Saldo Devedor (R$)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

# Totais
print(f"Valor à vista (PV): R$ {PV:.2f}")
print(f"Parcela fixa (incluindo entrada): R$ {P:.2f}")
print(f"Total pago: R$ {df['Pagamento (R$)'].sum():.2f}")
print(f"Total de juros pagos: R$ {df['Juros (R$)'].sum():.2f}")
```
